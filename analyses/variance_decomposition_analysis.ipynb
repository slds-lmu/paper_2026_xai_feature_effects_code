# Variance Decomposition Analysis

This notebook contains the analysis of the variance decomposition decomposition results (§6.2 in the paper).

```python
from configparser import ConfigParser
from pathlib import Path
import pandas as pd

from current_research_feature_effects.plotting.table_formatting import format_var_decomp_dataframe, highlight_min_feature_metric
```

```python
EXPERIMENT_PATH = Path("../experiments/main_study")
config = ConfigParser()
config.read(EXPERIMENT_PATH / "main_study.ini")
effects_results_storage = config.get("storage", "effects_results")
```

## SimpleNormalCorrelated

### PDP

```python
df_snc_pdp = pd.read_sql_table(
    "pdp_var_results", f"sqlite:///{EXPERIMENT_PATH}/SimpleNormalCorrelated/{effects_results_storage}"
)
format_var_decomp_dataframe(df_snc_pdp).style.apply(highlight_min_feature_metric, axis=None).format("{:.4f}")
```

## ALE

```python
df_snc_ale = pd.read_sql_table(
    "ale_var_results", f"sqlite:///{EXPERIMENT_PATH}/SimpleNormalCorrelated/{effects_results_storage}"
)
format_var_decomp_dataframe(df_snc_ale).style.apply(highlight_min_feature_metric, axis=None).format("{:.4f}")
```

```python
# plot_variance_table(
#     df_snc_ale,
#     models=["XGBoost_OT", "XGBoost_OF"],
#     type="ale",
#     # save_figs=Path("../paper/img/SNC-all/"),
#     # large_font=True,
#     # show_title=False,
# )
```

```python
# ---- for paper ----
feat = ["x_1", "x_2", "x_3"]
df_pdp_snc_ = df_snc_pdp.loc[df_snc_pdp["feature"].isin(feat)]

df_ale_snc_ = df_snc_ale.loc[df_snc_ale["feature"].isin(feat)]
out = pd.concat(
    {"PD": df_pdp_snc_, "ALE": df_ale_snc_},
    names=["effect"]
).reset_index()

format_var_decomp_dataframe(out, multiple_effects=True).style.apply(highlight_min_feature_metric, axis=None).format("{:.4f}").to_latex("tab/snc_var_decomp_results.tex", convert_css=True)
```
